# Text Preprocessing Techniques in NLP

This notebook demonstrates common text preprocessing steps in Natural Language Processing (NLP), including cleaning, tokenization, stop word removal, stemming, and lemmatization. Each step is explained simply and shown with code and output.

## 1. Import Required Libraries

Let's import the necessary Python libraries for text preprocessing. We'll use `re` for regular expressions and `nltk` for NLP tasks. We'll also download the required NLTK resources.

```python
import re
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Download required NLTK resources
nltk.download('punkt')  # Used for tokenization. In Windows, you might have to run `nltk.download('punkt_tab')` as well
nltk.download('stopwords')  # Used for removing stopwords
nltk.download('wordnet')  # Used for lemmatization
```

## 2. Sample Text Data

Let's define a sample text string to use for all preprocessing steps.

```python
# Example text for preprocessing
test_text = "NLP is fun! Let's clean, tokenize, and analyze this text: 123 times."
print(test_text)
```

## 3. Removing Unwanted Characters

We can use regular expressions to remove unwanted characters, such as punctuation, from the text. Adjust the regex pattern as needed for your use case.

```python
# Remove punctuation (keep only words and spaces)
cleaned_text = re.sub(r'[^\w\s]', '', test_text)
print(cleaned_text)

# Remove digits as well (keep only letters and spaces)
letters_only = re.sub(r'[^a-zA-Z\s]', '', test_text)
print(letters_only)
```

## 4. Lowercasing Text

Converting all text to lowercase helps ensure uniformity and reduces the vocabulary size.

```python
# Convert to lowercase
lower_text = cleaned_text.lower()
print(lower_text)
```

## 5. Tokenization

Tokenization splits the text into individual words or tokens. We'll use NLTK's `word_tokenize` function.

```python
# Tokenize the text
# (word_tokenize expects a string, so we use the lowercase cleaned text)
tokens = word_tokenize(lower_text)
print(tokens)
```

## 6. Removing Stop Words

Stop words are common words (like "and", "the", "is") that are often removed from text data because they carry little meaning.

```python
# Remove English stop words
stop_words = set(stopwords.words('english'))
filtered_tokens = [word for word in tokens if word not in stop_words]
print(filtered_tokens)
```

## 7. Stemming

Stemming reduces words to their root form by chopping off word endings. We'll use NLTK's `PorterStemmer`.

```python
# Apply stemming to filtered tokens
stemmer = PorterStemmer()
stemmed_tokens = [stemmer.stem(word) for word in filtered_tokens]
print(stemmed_tokens)
```

## 8. Lemmatization

Lemmatization reduces words to their base or dictionary form (lemma), considering the context. We'll use NLTK's `WordNetLemmatizer`.

```python
# Apply lemmatization to filtered tokens
lemmatizer = WordNetLemmatizer()
lemmatized_tokens = [lemmatizer.lemmatize(word) for word in filtered_tokens]
print(lemmatized_tokens)
```

> Note: You can observe the differences between stemming and lemmatization in the output.
> 
> - Stemming output for `tokenize` is `token`, while lemmatization output is `tokenize`.
> - Similarly, stemming `analyze` results in `analyz`, while lemmatization gives `analyze`.

# Natural Language Processing (NLP)

Natural Language Processing (NLP) is a subfield of artificial intelligence that focuses on the interaction between computers and human (natural) languages. The ultimate objective of NLP is to enable computers to understand, interpret, and generate human language in a way that is valuable.

## Key Concepts in NLP

1. **Tokenization**: The process of breaking down text into smaller units, such as words or phrases, called tokens.
2. **Part-of-Speech Tagging**: Assigning grammatical categories (like noun, verb, adjective) to each token in a sentence.
3. **Named Entity Recognition (NER)**: Identifying and classifying named entities (like people, organizations, locations) in text.
4. **Sentiment Analysis**: Determining the sentiment or emotional tone behind a body of text.
5. **Language Modeling**: Predicting the next word or sequence of words in a sentence based on the context of the preceding words.
6. **Machine Translation**: Automatically translating text from one language to another.
7. **Text Summarization**: Creating a concise summary of a longer text while retaining its key information.
8. **Word Embeddings**: Representing words in a continuous vector space where semantically similar words are mapped to nearby points.

## Common terminologies of NLP

- **Corpus**: A large and structured set of texts used for training and evaluating NLP models.
- **Document**: A single piece of text, such as an article, email, or book, that is part of a corpus.
- **Vocabulary**: The set of unique words or tokens present in a corpus or document.
- **Token**: A single unit of text, such as a word or punctuation mark, obtained through tokenization.
- **N-grams**: Contiguous sequences of 'n' items (words or characters) from a given text, used for modeling language.
- **Stop Words**: Commonly used words (like "and", "the", "is") that are often removed from text during preprocessing.
- **Stemming**: The process of reducing words to their root form (e.g., "running" to "run").
- **Lemmatization**: Similar to stemming, but it reduces words to their base or dictionary form (e.g., "better" to "good").
- **Bag of Words (BoW)**: A representation of text that describes the occurrence of words within a document, disregarding grammar and word order.
- **TF-IDF (Term Frequency-Inverse Document Frequency)**: A statistical measure used to evaluate the importance of a word in a document relative to a collection of documents (corpus).
- **Attention Mechanism**: A technique used in neural networks to focus on specific parts of the input sequence when making predictions, improving performance in tasks like translation and summarization.
- **Transformer**: A deep learning model architecture that uses self-attention mechanisms, widely used in NLP tasks for its efficiency and effectiveness.
- **BERT (Bidirectional Encoder Representations from Transformers)**: A pre-trained transformer model designed to understand the context of words in a sentence by looking at both directions (left and right).
- **GPT (Generative Pre-trained Transformer)**: A series of transformer-based models designed for generating human-like text based on the input provided.
- **Sequence-to-Sequence (Seq2Seq)**: A model architecture used for tasks where input sequences are transformed into output sequences, such as in machine translation.


## Text Preprocessing Steps

1. **Lowercasing**: Converting all text to lowercase to ensure uniformity.
2. **Removing Punctuation**: Eliminating punctuation marks to focus on words.
3. **Removing Stop Words**: Filtering out common words that do not add significant meaning.
4. **Stemming/Lemmatization**: Reducing words to their root or base form.
5. **Tokenization**: Splitting text into individual words or tokens.

## Text Representation Techniques

### One Hot Encoding

In the context of NLP, One Hot Encoding is a method used to represent words or tokens as binary vectors. Each word in the vocabulary is represented by a vector of length equal to the size of the vocabulary, where all elements are 0 except for the index corresponding to the word, which is set to 1.

**Example**:

Consider a vocabulary with the words: ["cat", "dog", "fish"]

| Word | One Hot Encoding      |
|------|----------------------|
| cat  | [1, 0, 0]            |
| dog  | [0, 1, 0]            |
| fish | [0, 0, 1]            |

**Pros**:
- Simple and easy to implement.
- Intuitive
- Works well for small vocabularies.

**Cons**:
- Sparse matrix representation.
- High dimensionality for large vocabularies.
- Out of vocabulary (OOV) issues with new words.
- Length of vectors increases with vocabulary size.
- Does not capture semantic relationships between words.

### Bag of Words (BoW)

Bag of Words (BoW) is a text representation technique that converts text into a fixed-length vector by counting the occurrences of each word in the vocabulary, disregarding grammar and word order.
BOW is a feature extraction technique used in natural language processing and information retrieval.

**Example**:

Consider the following two sentences:
1. "I love programming"
2. "Programming is fun"

| Word        | Sentence 1 Count | Sentence 2 Count |
|-------------|------------------|------------------|
| I           | 1                | 0                |
| love        | 1                | 0                |
| programming | 1                | 1                |
| is          | 0                | 1                |
| fun         | 0                | 1                |

The Bag of Words representation for the two sentences would be:
- Sentence 1: [1, 1, 1, 0, 0]
- Sentence 2: [0, 0, 1, 1, 1]

**Pros**:
- Simple and easy to implement.
- Captures word frequency information.
- Works well for text classification tasks.

**Cons**:
- Ignores word order and context.
- High dimensionality for large vocabularies.
- Out of vocabulary (OOV) issues with new words.
- Sparse representation.



### Term Frequency-Inverse Document Frequency (TF-IDF)

Term Frequency-Inverse Document Frequency (TF-IDF) is a statistical measure used to evaluate the importance of a word in a document relative to a collection of documents (corpus). It combines two metrics: Term Frequency (TF) and Inverse Document Frequency (IDF).

**Example**:

Consider the following two documents:
1. "I love programming"
2. "Programming is fun"

**Term Frequency (TF)**
TF(word) = (Number of times word appears in a document) / (Total number of words in the document)

**Inverse Document Frequency (IDF)**
IDF(word) = log_e(Total number of documents / Number of documents with the word in it)

**TF-IDF Calculation**
TF-IDF(word) = TF(word) * IDF(word)

**Example Calculation**:
For the word "programming" in Document 1:
- TF("programming") = 1/3 = 0.33
- IDF("programming") = log_e(2/2) = 0
- TF-IDF("programming") = 0.33 * 0 = 0

**Pros**:
- Captures word importance in context.

**Cons**:
- Ignores word order and context.
- High dimensionality for large vocabularies.
- Out of vocabulary (OOV) issues with new words.
- Sparse representation.

### N-grams

N-grams are contiguous sequences of 'n' items (words or characters) from a given text. In NLP, n-grams are commonly used to capture context and relationships between words in a sequence.

**Example**:

Consider the sentence: "I love programming"
- Unigrams (1-grams): ["I", "love", "programming"]
- Bigrams (2-grams): ["I love", "love programming"]
- Trigrams (3-grams): ["I love programming"]

### Text Preprocessing

#### Build corpus with removing all characters except alphabets, converting to lowercase

````python
import re
````

````python
corpus = []

for i in range(len(sentences)):
    review = re.sub('[^a-zA-Z]', ' ', sentences[i])
    review = review.lower()
    corpus.append(review)
corpus
````

````python
nltk.download('stopwords')
````

#### Removing Stopwords and then Stemming

````python
## Stemming
for i in corpus:
    words = nltk.word_tokenize(i)
    stemmed_words = [stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    print(" ".join(stemmed_words))
````

#### Removing Stopwords and then Lemmatization

````python
## Lemmatization
for i in corpus:
    words = nltk.word_tokenize(i)
    lemmatized_words = [lemmatizer.lemmatize(word) for word in words if word not in set(stopwords.words('english'))]
    print(" ".join(lemmatized_words))
````

#### Actual Text Preprocessing and building corpus

````python
corpus = []

for i in range(len(sentences)):
    review = re.sub('[^a-zA-Z]', ' ', sentences[i])
    review = review.lower()
    review = review.split()
    review = [lemmatizer.lemmatize(word) for word in review if word not in set(stopwords.words('english'))]
    review = ' '.join(review)
    corpus.append(review)
corpus
````
